```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
# Setting the working directory
import os
os.chdir(r'd:\Dowry-Demand-Prediction')
```

```python
df = pd.read_excel('Dataset\\cleaned\\Dowry-Data-Bangladesh-cleaned.xlsx')
```

##  Fix Skewed Numerical Features

```python
df['dowry_money_log'] = np.log1p(df['dowry_money'])
df["mohor_log"] = np.log1p(df["mohor"])
```

```python
df.drop(columns=['dowry_money', 'mohor'], inplace=True)
```

## Label Encoding

```python
# Columns to one-hot encode
categorical_cols = [
    'family type', 'area', "girl's job", "boy's job",
    'marry condition', 'women married/unmarried'
]

# One-hot encode the categorical columns
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True, dtype='int')
```

```python
# Label encode the target variable 'dowry'
le = LabelEncoder()
df_encoded['dowry'] = le.fit_transform(df['dowry'])
```

```python
# (Optional) Save the label encoder for inverse transform later
import joblib
joblib.dump(le, 'models/dowry_label_encoder.pkl')
```

```python
# Save the encoded data to the specified directory
encoded_output_file = "Dataset/encoded/Dowry-Data-Bangladesh-encoded.xlsx"
df_encoded.to_excel(encoded_output_file, index=False)
print(f"encoded data saved to {encoded_output_file}")
```

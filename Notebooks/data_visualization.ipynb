```python
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
import pandas as pd

# Setting the working directory
import os
os.chdir(r'd:\Dowry-Demand-Prediction')
```

```python
df = pd.read_excel('Dataset\\cleaned\\Dowry-Data-Bangladesh-cleaned.xlsx')
```

```python
def countplot(df,column):
    bar = sns.countplot(data=df,x=column,palette='rocket')
    for bars in bar.containers:
        bar.bar_label(bars)
    plt.tight_layout()
```

```python
plt.figure(figsize=(14,4))
countplot(df,"family type")
```

```python
plt.figure(figsize=(12,4))
countplot(df,"girl's job")
```

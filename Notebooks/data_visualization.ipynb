```python
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
import pandas as pd
import matplotlib.pyplot as plt
# Setting the working directory
import os
os.chdir(r'd:\Dowry-Demand-Prediction')
```

```python
df = pd.read_excel('Dataset\\cleaned\\Dowry-Data-Bangladesh-cleaned.xlsx')
```

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
```

```python
plt.figure(figsize=(15, 5 * ((len(df.columns) + 1) // 2)))

for i, column in enumerate(df.columns):
    plt.subplot((len(df.columns) + 1) // 2, 2, i + 1)
    if column in categorical_cols:
        sns.countplot(x=column, data=df, palette='Set2')
    else:
        sns.histplot(df[column], kde=True, color='skyblue')
    plt.title(f'Distribution of {column}')

plt.tight_layout()
plt.show()
```

```python
def countplot(df,column):
    bar = sns.countplot(data=df,x=column,palette='rocket')
    for bars in bar.containers:
        bar.bar_label(bars)
    plt.tight_layout()
```

```python
plt.figure(figsize=(14,4))
countplot(df,"family type")
```
